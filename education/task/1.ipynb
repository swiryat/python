```python
# изучаем содержимое строки и обращаем внимание,
# что перед началом каждом задачи идет одна и также последовательность символов `\r\n\r\n`
# воспользуемся методом split() и разделим строку по этой последовательности
lst_task = task.split('\r\n\r\n')
len(lst_task)
```

```python
#Убираем пустые задачи вида ''
for t in lst_task: # перебираем последовательно список задач
    if len(t.strip()) == 0:
       lst_task.remove(t) #то мы нашли пустую задачу (строку), удаляем ее из списка
```

```python
new_lst_task = list()
for t in lst_task: # перебираем последовательно список задач
    if t.split('.')[0].isdigit(): #пользуемся методом isdigit() -- проверяет, является ли строка числом
        new_lst_task.append(t) #добавляем в новый список
```

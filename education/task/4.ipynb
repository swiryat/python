````python
def median(lst: list) -> float:
    lst1 = sorted(lst)
    lst.sort()
    if len(lst) % 2 == 0:
        return (lst[int(len(lst)/2) -1] + lst[int(len(lst)/2)])/2
    else:
        return lst[int((len(lst) + 1)/2) -1]
````

````python
def mode(lst: list) -> list:
    """
    Функция для расчета моды величины.
    Напомним, мода это значение(я) величины, которое
    встречается чаще всего.
    Например, для списка [1,2,2,3,3] -- мода 2 и 3
    для списка [1,1,1,1,2] -- мода 1.
    """
    # алгоритм работы функции таков
    # 1. находим уникальные значения -- с помощью конвертации во множество;
    # 2. создаем вспомотельный словарь с ключами -- уникальными значениями;
    # 3. считаем сколько раз встречалось каждое из уникальных значений в исходном списке;
    # 4. находим максимальное количество повторов -- выводим ключ.

    mode_list = list() #список для значений мод;
    dict_value_list = dict.fromkeys(set(lst), None) # вспомогательный словарь с ключами -- уникальными значениями входного списка

    for key in dict_value_list.keys():
        dict_value_list[key] = lst.count(key) #считаем сколько раз каждый ключ встретился в исходном списке и пишем в словарь

    for key, value in dict_value_list.items(): # перебираем ключи и значения - частоту каждого значения
        if value == max(dict_value_list.values()): # если текущая частота равна макимальной, то
            mode_list.append(key) # мода и добавляем ее в список

    return mode_list
````

**Бизнес-метрики**

Необходимость в использовании аналитики:
- определить целевую аудиторию клиентов;
- понимать состояние компании;
- формулировать цели/строить прогнозы;
- использовать имеющиеся данные/информацию для достижения целей компании (получение прибыли);
- принимать эффективные решения;
- составлять план действий.

Задачи аналитики:
- прогнозировать сценарии развития;
- определять точки роста и успешные действия;
- выявить потребности пользователей;
- анализировать внешнюю среду (по рынку, по конкурентам);
- предоставлять информацию для стейкхолдеров.

[Принятие решений на данных](https://vc.ru/growth/621373-bitva-kultur-prinyatiya-resheniy-driven-vs-informed-vs-inspired)

Бизнес-метрики --- показатели, с помощью которых измеряют эффективность развития бизнеса на каждом этапе.

Функции метрик:
- Количественное отображение бизнес-процессов;
- Оперативная реакция на негативные/позитивные изменения;
- Определение ключевых точек в ЖЦ продукта (рост/падение спроса);
- Мониторинг динамики изменений в каждом периоде ЖЦ продукта (эффективность рекламы, каналов клиентов, качество продукта).

Выделяют следующие виды метрик:
- продуктовые;
- метрики электронной коммерции;
- маркетинговые.

Продуктовые метрики — это количественно измеримые показатели, которые компания отслеживает и анализирует для оценки успеха своего продукта.

|Продуктовые метрики|
|:--|
|Средний доход на пользователя (ARPU) -- это доход, получаемый (в среднем) на каждого активного пользователя вашего приложения|
|Пожизненная ценность клиента (LTV) -- ожидаемый валовый доход, который приносит клиент за все время использования продукта|
|Количество уникальных пользователей в день (DAU)|
|Количество уникальных пользователей в неделю (WAU)|
|Количество уникальных пользователей в месяц (MAU)|
|Коэффициент удержания клиентов (Retention Rate) -- процент клиентов, которых бизнес удерживает в течение определенного периода времени|

Показатели электронной коммерции -- любые поддающиеся количественной оценке, последовательно определенные измерения эффективности веб-сайта онлайн-бизнеса.

|Метрики электронной коммерции|
|:--|
|Коэффициент конверсии продаж -- это процент, который определяет количество клиентов, совершивших покупку|
|Темп -- количество пользователей, регулярно посещающих ресурс|
|Показатель отказов -- это процент, используемый для анализа посещаемости веб-сайта|
|Среднее время на странице -- среднее количество времени, затрачиваемое на одну веб-страницу всеми пользователями ресурса|
|Ценность страницы -- cредняя ценность страницы, которую посетил пользователь перед переходом на целевую страницу или завершением транзакции электронной торговли|

**Маркетинговые метрики**

Маркетинговые метрики -- метрики, отражающие эффективность маркетинговых и рекламных компаний.

|*Маркетинговые метрики*|
|:--|
|Коэффициент конверсии (CR)|
|Кликабельность (CTR)|
|Цена за один клик по рекламе(CPC)|
|Цена за одно целевое действие (CPA)|
|Стоимость привлечения клиента (CAC)|

**Когортный анализ**

Когорта группа людей, объединённых одним и тем же событием, произошедшим в определённый период времени.

Когортный анализ -- метод маркетингового исследования поведения пользователей, объединенных в когорты (группы) на основании действий, выполненных за определенный период. Этот метод позволяет подобрать ключевые метрики для каждой аудитории и увидеть эффективность продвижения бизнеса и маркетинговых кампаний.

Когортный анализ позволяет исследовать изменения в поведении групп пользователей с течением времени и находить закономерности. Когортный анализ повзоляет:
- проанализировать влияние принятых маркетинговых решений на уровень удержания, ROI, LTV, CAC и другие метрики;
- найти наиболее эффективные каналы привлечения покупателей;
- оценить результативность рекламных кампаний;
- найти пути повышения вовлеченности пользователей с низкой активностью и минимизировать отток клиентов;
- увидеть необходимость подключения нового канала продвижения;
- повысить уровень конверсии;
- оценить влияние сезонности на уровень продаж;
- лучше узнать целевую аудиторию;
- эффективнее распределить маркетинговый бюджет.

**A/B тестирование**

[A/B-тестирование](https://ru.wikipedia.org/wiki/A/B-%D1%82%D0%B5%D1%81%D1%82%D0%B8%D1%80%D0%BE%D0%B2%D0%B0%D0%BD%D0%B8%D0%B5) -- метод маркетингового исследования, суть которого заключается в том, что контрольная группа элементов сравнивается с набором тестовых групп, в которых один или несколько показателей были изменены для того, чтобы выяснить, какие из изменений улучшают целевой показатель. Таким образом в ходе теста сравнивается вариант «A» и вариант «B», и целью является определение лучшего из двух протестированных вариантов.

Минимальный размер выборки, необходимый для детектирования относительного эффекта [[источник](https://habr.com/ru/companies/ru_mts/articles/700992/)]:

$$
n = \dfrac{(\sigma_x^2+\sigma_y^2)(\Phi^{-1}(1-\frac{\alpha}{2})-\Phi^{-1}(\beta))^2}{\mu_x\delta}.
$$

Здесь $X$ и $Y$ две выборки -- контрольная и тестовая; $\alpha$ и $\beta$ -- ошибки 1-го и 2-го рода; $\delta$ -- оценка относительного прироста показателя; $\Phi()$ -- функция Лапласа.

```python
# Распределение событий: какие события и в каком количестве.
funnel_246 = df_ab[df.grop_id == 246].groupby(['grop_id', 'event_name'],as_index=False)\
         .agg(count_events = ('event_name', 'count'))\
         .sort_values(by='count_events', ascending=False)

funnel_246

funnel_248 = df_ab[df.grop_id == 248].groupby(['grop_id', 'event_name'],as_index=False)\
         .agg(count_events = ('event_name', 'count'))\
         .sort_values(by='count_events', ascending=False)

funnel_248


funnel_full_inner = funnel_246.merge(funnel_248,
                                     on='event_name',
                                     suffixes=('_A', '_B'),
                                     how=)
funnel_full_inner

funnel_full = funnel_246.append(funnel_248)
funnel_full
funnel_full_inner
```

```python
#Сколько пользователей совершали каждое из этих событий?

df_ab.groupby(['grop_id', 'event_name'], as_index=False)\
         .agg(count_users = ('user_id', 'nunique'))#\
         #.sort_values(by='count_events', ascending=False)

funnel_ab = pd.pivot_table(data=df_ab,
               index='event_name',
               columns='grop_id',
               #values='user_id',
               aggfunc={'user_id': 'nunique'})\
               .reset_index()
funnel_ab
funnel_ab.columns = ['event_name', 'count_users_246', 'count_users_248']
funnel_ab = funnel_ab.sort_values(by='count_users_246', ascending = False)
funnel_ab
```

```python
# Проверьте гипотезу о наличие значимых отличий по результатам теста.

# H0: нет статистически значимой разницы между долями пользователей из обеих групп, совершивших одно и тоже событие
# H1: имеется статистически значимая разница между долями пользователей из обеих групп, совершивших одно и тоже событие
```

Для биномиальных распределений

$$
\begin{multline*}
z = \dfrac{\dfrac{m_1}{n_1}+ \dfrac{1}{2n_1} - \dfrac{m_2}{n_2} - \dfrac{1}{2n_2}}{\sqrt{\dfrac{m_1+m_2}{n_1+n_2} \dfrac{n_1+n_2 - m_1 - m_2}{n_1+n_2}\left(\dfrac{1}{n_1}+\dfrac{1}{n_2}\right)}} = \\
= \dfrac{\dfrac{m_1}{n_1}- \dfrac{m_2}{n_2} + \dfrac{1}{2n_1} - \dfrac{1}{2n_2}}{\sqrt{\dfrac{m_1+m_2}{n_1+n_2} \left( 1-\dfrac{m_1+m_2}{n_1+n_2}\right)\left(\dfrac{1}{n_1}+\dfrac{1}{n_2}\right)}}
\end{multline*}
$$*Курсив*

n1 -- количество уникальных клиентов группы A
m1 -- количество уникальных клиентов группы A, совершивших некоторое событие


n2 -- количество уникальных клиентов группы B
m2 -- количество уникальных клиентов группы B, совершивших некоторое событие


```python
funnel_ab.index = funnel_ab['event_name']
funnel_ab;
```

```python
#для биноминальных распределений
def test_z_criterion(event_name, alpha):
    import scipy.stats as st
    import math
    n1 = df_ab[df_ab['grop_id'] == 246].user_id.nunique()
    n2 = df_ab[df_ab['grop_id'] == 248].user_id.nunique()

    m1 = funnel_ab.loc[event_name, 'count_users_246']
    m2 = funnel_ab.loc[event_name, 'count_users_248']

    z_value = (m1/n1 - m2/n2) / math.sqrt(((m1+m2)/(n1+n2))*(1-((m1+m2)/(n1+n2)))*(1/n1+1/n2))
    distr = st.norm(0, 1)
    p_value = (1 - distr.cdf(abs(z_value))) * 2 #1-F(t)
    print('Результаты проверки гипотезы H_0 по z-критерию:')
    print('Уровень значимости alpha={:.2f}'.format(alpha))
    print('P-value: {:.2f}'.format(p_value))
    if p_value < alpha:
        print(f"""Отвергаем нулевую гипотезу: разница в долях
по событию <<{event_name}>> статистически значима.""")
    else: print(f"""Не получилось отвергнуть нулевую гипотезу: разница в долях по
событию <<{event_name}>> статистически незначима.""")
    print('------------------------------------------------------------')
```

```python
alpha = 0.05
for e in list(funnel_ab['event_name'].unique()):
    test_z_criterion(e, alpha)
```
